```python
lista = [10, 50, 150, 20, 300]
```

```python
def valores(lista: list[int]) -> None:
    for valores in lista:
        print(valores)
valores(lista)
```

```python
def alerta(lista: list[int]) -> list[int]:
    lista_alerta = []

    for valores in lista:
        if valores > 100:
            lista_alerta.append(valores)
    return lista_alerta

resultado = alerta(lista)

print(resultado)
```

```python
if resultado == []:
    print("Sistema Seguro")
else:
    print(f"Revisar valores: {resultado}")
```
